```python
# ============================================================
# Celda 1: Instalación e imports
# ============================================================
!pip install "gymnasium[atari]" ale-py matplotlib numpy torch pillow -q

import base64
import random
import time
import gc

import numpy as np
import matplotlib.pyplot as plt
import gymnasium as gym
import ale_py
import torch
import torch.nn as nn
import torch.nn.functional as F
from PIL import Image
from IPython.display import HTML, display

try:
    from gymnasium.wrappers import AtariPreprocessing, FrameStackObservation as FrameStack
except ImportError:  # versiones anteriores de gymnasium
    from gymnasium.wrappers import AtariPreprocessing, FrameStack

if hasattr(gym, "register_envs"):
    gym.register_envs(ale_py)

SEED = 0
random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
torch.backends.cudnn.benchmark = True
print("Dispositivo:", device)

# ============================================================
# Celda 2: Entorno (Enduro) y parámetros
# ============================================================
ENV_ID = "ALE/Enduro-v5"

def make_env(render=False):
    """Enduro con el preprocesamiento clásico de DQN: frame skip 4, 84x84 en grises, 4 cuadros apilados."""
    env = gym.make(
        ENV_ID,
        frameskip=1,                    # el salto de cuadros lo hace AtariPreprocessing
        repeat_action_probability=0.0,  # sin 'sticky actions', como en el paper original
        full_action_space=False,
        render_mode="rgb_array" if render else None,
    )
    env = AtariPreprocessing(env, noop_max=30, frame_skip=4, screen_size=84,
                             terminal_on_life_loss=False, grayscale_obs=True, scale_obs=False)
    env = FrameStack(env, 4)
    return env

_tmp_env = make_env()
N_ACTIONS = _tmp_env.action_space.n          # 9 acciones en Enduro
print("Cantidad de acciones:", N_ACTIONS)
_tmp_env.close()

# --- Hiperparámetros (configuración inicial, deliberadamente deficiente) ---
NUM_STEPS = 100_000          # pasos de entorno por agente (cada paso = 4 cuadros del juego)
GAMMA = 0.99
LR = 1e-3
BATCH_SIZE = 32
BUFFER_SIZE = 1_000          # capacidad del replay buffer
LEARN_START = 1_000          # pasos de experiencia antes de empezar a aprender
TRAIN_EVERY = 4              # una actualización de la red cada 4 pasos de entorno
TARGET_UPDATE_EVERY = 1      # pasos entre cada copia de la red principal a la red objetivo
EPSILON_START = 1.0
EPSILON_END = 0.01
EPSILON_DECAY_STEPS = 5_000  # pasos durante los cuales epsilon decae linealmente

# ============================================================
# Celda 3: Replay buffer (almacena los estados como uint8)
# ============================================================
class ReplayBuffer:
    def __init__(self, capacity):
        self.capacity = capacity
        self.pos = 0
        self.size = 0
        self.s = np.zeros((capacity, 4, 84, 84), dtype=np.uint8)
        self.s2 = np.zeros((capacity, 4, 84, 84), dtype=np.uint8)
        self.a = np.zeros(capacity, dtype=np.int64)
        self.r = np.zeros(capacity, dtype=np.float32)
        self.d = np.zeros(capacity, dtype=np.float32)

    def push(self, s, a, r, s2, done):
        self.s[self.pos] = s
        self.s2[self.pos] = s2
        self.a[self.pos] = a
        self.r[self.pos] = r
        self.d[self.pos] = done
        self.pos = (self.pos + 1) % self.capacity
        self.size = min(self.size + 1, self.capacity)

    def sample(self, batch_size):
        idx = np.random.randint(0, self.size, size=batch_size)
        return (
            torch.as_tensor(self.s[idx], device=device),
            torch.as_tensor(self.a[idx], device=device).unsqueeze(1),
            torch.as_tensor(self.r[idx], device=device).unsqueeze(1),
            torch.as_tensor(self.s2[idx], device=device),
            torch.as_tensor(self.d[idx], device=device).unsqueeze(1),
        )

    def __len__(self):
        return self.size

# ============================================================
# Celda 4: Redes neuronales (DQN clásica y Dueling DQN)
# ============================================================
class ConvTrunk(nn.Module):
    """Tronco convolucional común (arquitectura del paper de DQN en Nature)."""
    def __init__(self):
        super().__init__()
        self.conv = nn.Sequential(
            nn.Conv2d(4, 32, kernel_size=8, stride=4), nn.ReLU(),
            nn.Conv2d(32, 64, kernel_size=4, stride=2), nn.ReLU(),
            nn.Conv2d(64, 64, kernel_size=3, stride=1), nn.ReLU(),
            nn.Flatten(),
        )
        self.out_dim = 64 * 7 * 7   # 3136

    def forward(self, x):
        return self.conv(x.float() / 255.0)


class DQN(nn.Module):
    """Red clásica: una sola cabeza que devuelve Q(s, a) para cada acción."""
    def __init__(self, n_actions):
        super().__init__()
        self.trunk = ConvTrunk()
        self.fc = nn.Sequential(nn.Linear(self.trunk.out_dim, 512), nn.ReLU(),
                                nn.Linear(512, n_actions))

    def forward(self, x):
        return self.fc(self.trunk(x))


class DuelingDQN(nn.Module):
    """Red Dueling: mismo tronco, pero dos ramas (V(s) y A(s,a)) que se recombinan."""
    def __init__(self, n_actions):
        super().__init__()
        self.trunk = ConvTrunk()
        self.value = nn.Sequential(nn.Linear(self.trunk.out_dim, 512), nn.ReLU(),
                                   nn.Linear(512, 1))
        self.advantage = nn.Sequential(nn.Linear(self.trunk.out_dim, 512), nn.ReLU(),
                                       nn.Linear(512, n_actions))

    def forward(self, x):
        h = self.trunk(x)
        v = self.value(h)
        a = self.advantage(h)
        # Q(s,a) = V(s) + ( A(s,a) - media_a' A(s,a') )
        return v + a - a.mean(dim=1, keepdim=True)

# ============================================================
# Celda 5: Agente y entrenamiento (idéntico para ambas redes)
# ============================================================
def train_agent(network_class, seed=SEED, num_steps=None, verbose=True):
    num_steps = num_steps or NUM_STEPS
    env = make_env()
    random.seed(seed); np.random.seed(seed); torch.manual_seed(seed)

    policy_net = network_class(N_ACTIONS).to(device)
    target_net = network_class(N_ACTIONS).to(device)
    target_net.load_state_dict(policy_net.state_dict())
    optimizer = torch.optim.Adam(policy_net.parameters(), lr=LR)
    buffer = ReplayBuffer(BUFFER_SIZE)

    state, _ = env.reset(seed=seed)
    state = np.array(state, dtype=np.uint8)
    ep_return = 0.0
    log_steps, log_returns = [], []
    t0 = time.time()

    for step in range(1, num_steps + 1):
        # Política de comportamiento epsilon-greedy con decaimiento lineal
        epsilon = max(EPSILON_END,
                      EPSILON_START - (EPSILON_START - EPSILON_END) * step / EPSILON_DECAY_STEPS)
        if random.random() < epsilon:
            action = random.randrange(N_ACTIONS)
        else:
            with torch.no_grad():
                s_t = torch.as_tensor(state, device=device).unsqueeze(0)
                action = int(policy_net(s_t).argmax(dim=1).item())

        next_state, reward, terminated, truncated, _ = env.step(action)
        next_state = np.array(next_state, dtype=np.uint8)
        done = terminated or truncated
        # Solo 'terminated' corta el bootstrap (truncated no es un estado terminal real)
        buffer.push(state, action, reward, next_state, float(terminated))
        state = next_state
        ep_return += reward

        if len(buffer) >= max(BATCH_SIZE, LEARN_START) and step % TRAIN_EVERY == 0:
            s, a, r, s2, d = buffer.sample(BATCH_SIZE)
            q_sa = policy_net(s).gather(1, a)
            with torch.no_grad():
                q_next = target_net(s2).max(dim=1, keepdim=True)[0]
                target = r + GAMMA * q_next * (1.0 - d)
            loss = F.mse_loss(q_sa, target)
            optimizer.zero_grad()
            loss.backward()
            optimizer.step()

        if step % TARGET_UPDATE_EVERY == 0:
            target_net.load_state_dict(policy_net.state_dict())

        if done:
            log_steps.append(step)
            log_returns.append(ep_return)
            if verbose:
                print(f"  Paso {step:8d} | ε={epsilon:.3f} | Episodio {len(log_returns):3d} | "
                      f"Retorno: {ep_return:6.1f} | Media (últ. 5): {np.mean(log_returns[-5:]):6.1f}")
            ep_return = 0.0
            state, _ = env.reset()
            state = np.array(state, dtype=np.uint8)

    if verbose:
        print(f"  Tiempo de entrenamiento: {(time.time() - t0) / 60:.1f} min")
    env.close()
    del buffer
    gc.collect()
    if device.type == "cuda":
        torch.cuda.empty_cache()
    return policy_net, np.array(log_steps), np.array(log_returns)

# ============================================================
# Celda 6: Entrenamiento de ambos agentes
# ============================================================
print("Entrenando DQN clásica...")
dqn_net, steps_dqn, returns_dqn = train_agent(DQN)

print("\nEntrenando Dueling DQN...")
dueling_net, steps_dueling, returns_dueling = train_agent(DuelingDQN)

# ============================================================
# Celda 7: Visualización comparativa
# ============================================================
def moving_average(x, w=5):
    w = min(w, len(x))
    return np.convolve(x, np.ones(w) / w, mode='valid')

plt.figure(figsize=(10, 5))
for steps, rets, color, label in [(steps_dqn, returns_dqn, 'blue', 'DQN'),
                                  (steps_dueling, returns_dueling, 'red', 'Dueling DQN')]:
    ma = moving_average(rets)
    plt.plot(steps[len(steps) - len(ma):], ma, color=color, label=label)
plt.xlabel('Pasos de entorno')
plt.ylabel('Retorno por episodio (media móvil de 5)')
plt.title(f'DQN vs Dueling DQN en {ENV_ID}')
plt.legend()
plt.grid(alpha=0.3)
plt.show()

# ============================================================
# Celda 8: Evaluación de las políticas finales
# ============================================================
EVAL_EPSILON = 0.01   # pequeña exploración para evitar que el agente quede atascado

def evaluate(net, n_eval=3, max_steps=10_000):
    eval_env = make_env()
    scores = []
    for i in range(n_eval):
        state, _ = eval_env.reset(seed=10_000 + i)
        done, total, steps = False, 0.0, 0
        while not done and steps < max_steps:
            if random.random() < EVAL_EPSILON:
                action = random.randrange(N_ACTIONS)
            else:
                with torch.no_grad():
                    s_t = torch.as_tensor(np.array(state, dtype=np.uint8), device=device).unsqueeze(0)
                    action = int(net(s_t).argmax(dim=1).item())
            state, reward, terminated, truncated, _ = eval_env.step(action)
            done = terminated or truncated
            total += reward
            steps += 1
        scores.append(total)
    eval_env.close()
    return np.mean(scores), np.std(scores)

m_dqn, s_dqn = evaluate(dqn_net)
m_duel, s_duel = evaluate(dueling_net)
print(f"\n✅ DQN         → retorno medio: {m_dqn:7.1f} ± {s_dqn:.1f}")
print(f"✅ Dueling DQN → retorno medio: {m_duel:7.1f} ± {s_duel:.1f}")

# ============================================================
# Celda 9: Renderizado en GIF de cada agente
# ============================================================
def record_gif(net, filename, seed=123, max_steps=1500, frame_skip=3):
    """Ejecuta un episodio con la política del agente y guarda el video como GIF."""
    render_env = make_env(render=True)
    state, _ = render_env.reset(seed=seed)
    frames, total, steps = [], 0.0, 0
    done = False
    while not done and steps < max_steps:
        if steps % frame_skip == 0:
            frame = Image.fromarray(render_env.render()).resize((160, 105))
            frames.append(frame)
        if random.random() < EVAL_EPSILON:
            action = random.randrange(N_ACTIONS)
        else:
            with torch.no_grad():
                s_t = torch.as_tensor(np.array(state, dtype=np.uint8), device=device).unsqueeze(0)
                action = int(net(s_t).argmax(dim=1).item())
        state, reward, terminated, truncated, _ = render_env.step(action)
        done = terminated or truncated
        total += reward
        steps += 1
    render_env.close()
    frames[0].save(filename, save_all=True, append_images=frames[1:], duration=50, loop=0)
    return total, steps

ret_d, steps_d = record_gif(dqn_net, "dqn_enduro.gif")
ret_u, steps_u = record_gif(dueling_net, "dueling_dqn_enduro.gif")

def gif_b64(path):
    return base64.b64encode(open(path, "rb").read()).decode()

display(HTML(f"""
<div style="display:flex; gap:30px; align-items:flex-start;">
  <div style="text-align:center">
    <b>DQN</b><br>
    <img src="data:image/gif;base64,{gif_b64('dqn_enduro.gif')}" width="320"><br>
    Autos sobrepasados: {ret_d:.0f} | Pasos: {steps_d}
  </div>
  <div style="text-align:center">
    <b>Dueling DQN</b><br>
    <img src="data:image/gif;base64,{gif_b64('dueling_dqn_enduro.gif')}" width="320"><br>
    Autos sobrepasados: {ret_u:.0f} | Pasos: {steps_u}
  </div>
</div>
"""))
```

Tablas. En las tablas, sem k es la evaluación de la semilla k (media de 10 episodios), media ± ee es la media entre semillas y su error estándar, últ. 20 ep es el retorno medio de los últimos 20 episodios de entrenamiento, Q sondas el valor en los estados de referencia al final y min los minutos por corrida, con varias corriendo en paralelo. En Enduro el retorno suma 1 por cada auto que el agente pasa y resta 1 por cada auto que lo pasa a él. La política al azar da 0 en los 10 episodios de evaluación: acelera (FIRE) en 3 de cada 9 acciones y frena (DOWN) en otras 3, y no llega a pasar autos.

```python
# ============================================================
# Preparación: hardware, semillas, ejecución en paralelo y diagnósticos
# ============================================================
import os
import platform
import psutil
from joblib import Parallel, delayed
from joblib.externals.loky import get_reusable_executor

plt.style.use('dark_background')
COLORES = ['#3987e5', '#d95926', '#199e70', '#c98500', '#d55181', '#008300', '#9085e9', '#e66767']
ARQUITECTURAS = (DQN, DuelingDQN)
NOMBRE = {'DQN': 'DQN', 'DuelingDQN': 'Dueling DQN'}
COLOR_ARQ = {'DQN': COLORES[0], 'DuelingDQN': COLORES[1]}
SEMILLAS = [0, 1, 2]          # SEED = 0 es la semilla del código original
SEMILLAS_EXTRA = [3, 4, 5]    # pregunta 7: tres semillas más con la configuración final
N_EVAL = 10                   # episodios de evaluación; los 3 primeros son los de evaluate
HIPERPARAMETROS = ['NUM_STEPS', 'GAMMA', 'LR', 'BATCH_SIZE', 'BUFFER_SIZE', 'LEARN_START', 'TRAIN_EVERY',
                   'TARGET_UPDATE_EVERY', 'EPSILON_START', 'EPSILON_END', 'EPSILON_DECAY_STEPS',
                   'HUBER', 'MAX_GRAD_NORM']
_env = make_env()
ACCIONES = _env.unwrapped.get_action_meanings()
_env.close()
ACELERA = [i for i, n in enumerate(ACCIONES) if 'FIRE' in n]   # en Enduro FIRE es el acelerador
FRENA = [i for i, n in enumerate(ACCIONES) if 'DOWN' in n]
print("Acciones:", ', '.join(f"{i} {n}" for i, n in enumerate(ACCIONES)))


def nombre_cpu():
    """Nombre comercial del procesador (registro en Windows, /proc/cpuinfo en Linux)."""
    try:
        if platform.system() == 'Windows':
            import winreg
            clave = winreg.OpenKey(winreg.HKEY_LOCAL_MACHINE,
                                   r'HARDWARE\DESCRIPTION\System\CentralProcessor\0')
            return winreg.QueryValueEx(clave, 'ProcessorNameString')[0].strip()
        with open('/proc/cpuinfo') as f:
            for linea in f:
                if linea.startswith('model name'):
                    return linea.split(':', 1)[1].strip()
    except OSError:
        pass
    return platform.processor() or 'CPU'


def ms_por_llamada(fn, n):
    """Milisegundos por llamada, sincronizando la GPU para medir el tiempo real."""
    fn()  # calentamiento
    if torch.cuda.is_available():
        torch.cuda.synchronize()
    t0 = time.perf_counter()
    for _ in range(n):
        fn()
    if torch.cuda.is_available():
        torch.cuda.synchronize()
    return (time.perf_counter() - t0) / n * 1e3


def medir_dispositivo(dev, n=100):
    """Tiempo de elegir una acción (lote de 1) y de una actualización (lote de 32, red objetivo incluida)."""
    net, objetivo = DQN(N_ACTIONS).to(dev), DQN(N_ACTIONS).to(dev)
    opt = torch.optim.Adam(net.parameters(), lr=1e-4)
    s1 = torch.randint(0, 256, (1, 4, 84, 84), dtype=torch.uint8, device=dev)
    s32 = torch.randint(0, 256, (32, 4, 84, 84), dtype=torch.uint8, device=dev)

    def elegir():
        with torch.no_grad():
            int(net(s1).argmax(dim=1).item())

    def actualizar():
        with torch.no_grad():
            y = objetivo(s32).max(dim=1, keepdim=True)[0]
        loss = F.smooth_l1_loss(net(s32)[:, :1], y)
        opt.zero_grad()
        loss.backward()
        opt.step()
    return ms_por_llamada(elegir, n), ms_por_llamada(actualizar, n)


def ms_por_paso_entorno(n=1000):
    """Paso de Enduro con el preprocesamiento de make_env (siempre en CPU)."""
    env = make_env()
    env.reset(seed=0)
    t0 = time.perf_counter()
    for i in range(n):
        _, _, term, trunc, _ = env.step(i % N_ACTIONS)
        if term or trunc:
            env.reset()
    env.close()
    return (time.perf_counter() - t0) / n * 1e3


# Se prefiere GPU cuando el cálculo es tensorial: cada paso elige la acción con la red convolucional
# y cada 4 pasos hay una actualización con un lote de 32 estados de 4x84x84.
CPU = nombre_cpu()
GPU = torch.cuda.get_device_name(0) if torch.cuda.is_available() else None
print(f"CPU detectada: {CPU} ({os.cpu_count()} hilos), RAM libre {psutil.virtual_memory().available / 1e9:.1f} GB")
print(f"GPU detectada: {GPU or 'ninguna'}")
t_env = ms_por_paso_entorno()
print(f"Paso del entorno con el preprocesamiento (siempre en CPU): {t_env:.2f} ms")
tiempos = {'CPU': medir_dispositivo(torch.device('cpu'))}
if GPU:
    tiempos['GPU'] = medir_dispositivo(torch.device('cuda'))
for nombre_disp, (t_accion, t_update) in tiempos.items():
    print(f"{nombre_disp}: elegir la acción {t_accion:.2f} ms, actualización con lote de 32 {t_update:.2f} ms, "
          f"unos {1e3 / (t_env + t_accion + t_update / TRAIN_EVERY):.0f} pasos de entrenamiento por segundo")
print(f"Dispositivo seleccionado: {'GPU ' + GPU if GPU else 'CPU ' + CPU}")


# ============================================================
# Estados de referencia y diagnósticos que registra cada corrida
# ============================================================
def estados_aleatorios(n=512, seed=123):
    """Estados de dos episodios con acciones al azar (como en el paper de DQN, para seguir Q en estados fijos)."""
    rng = np.random.default_rng(seed)
    env = make_env()
    estados = []
    for ep in range(2):
        state, _ = env.reset(seed=seed + ep)
        done = False
        while not done:
            estados.append(np.array(state, dtype=np.uint8))
            state, _, term, trunc, _ = env.step(int(rng.integers(N_ACTIONS)))
            done = term or trunc
    env.close()
    idx = np.linspace(0, len(estados) - 1, n).astype(int)
    return np.stack([estados[i] for i in idx])


SONDAS = estados_aleatorios()


class Registro:
    """Diagnósticos de una corrida, sin tocar los números al azar del entrenamiento:
    por episodio (retorno, autos pasados, acciones), cada LOG_CADA pasos (pérdida, error TD, norma
    del gradiente), cada 20 actualizaciones (cuánto se mueve el objetivo calculado con la propia red)
    y cada SONDA_CADA pasos (Q en los estados de referencia y un lote típico del buffer)."""
    LOG_CADA, SONDA_CADA, BLANCO_CADA = 1000, 5000, 20

    def __init__(self, net, buffer, seed):
        self.net, self.buffer = net, buffer
        self.rng = np.random.default_rng(seed + 1000)   # generador propio para el lote típico
        self.sondas = torch.as_tensor(np.array(SONDAS), device=device)   # copia escribible
        self.acciones = np.zeros(N_ACTIONS, dtype=int)
        self.pasados = self.sobrepasado = 0
        self.inicio_ep, self.fin_eps = 1, []
        self.n_upd = 0
        self.acum = torch.zeros(6, device=device)
        self.n_acum = 0
        self.d = {k: [] for k in ['ep_paso', 'ep_retorno', 'ep_epsilon', 'ep_pasados', 'ep_sobrepasado',
                                  'ep_acciones', 'log_paso', 'perdida', 'td_abs', 'td_mayor_1', 'grad',
                                  'grad_recorte', 'q_sa', 'blanco_paso', 'dq', 'dy', 'sonda_paso', 'q_max',
                                  'v', 'disp_a', 'lote_paso', 'lote_edad', 'lote_episodios', 'lote_mismo_ep',
                                  'lote_corr']}

    def paso(self, action, reward):
        self.acciones[action] += 1
        self.pasados += reward > 0
        self.sobrepasado += reward < 0

    def antes_del_paso(self, loss, q_sa, target, s2, d):
        """Después de backward y antes de optimizer.step: error TD y norma del gradiente (sin recortar)."""
        self.n_upd += 1
        td = (target - q_sa).detach()
        norma = nn.utils.get_total_norm([p.grad for p in self.net.parameters()])
        self.acum += torch.stack([loss.detach(), td.abs().mean(), (td.abs() > 1).float().mean(), norma,
                                  (norma > 10).float(), q_sa.detach().mean()])
        self.n_acum += 1
        self._medir_blanco = self.n_upd % self.BLANCO_CADA == 0
        if self._medir_blanco:
            with torch.no_grad():
                self._q_antes = q_sa.detach().clone()
                self._max_s2_antes = self.net(s2).max(dim=1, keepdim=True)[0]

    def despues_del_paso(self, step, s, a, s2, d):
        """Cuánto subió Q(s, a) con la actualización y cuánto subió, por arrastre, r + γ max Q(s′, ·)
        calculado con la misma red: es el objetivo que se usaría sin red objetivo."""
        if not self._medir_blanco:
            return
        with torch.no_grad():
            dq = self.net(s).gather(1, a) - self._q_antes
            dy = GAMMA * (1.0 - d) * (self.net(s2).max(dim=1, keepdim=True)[0] - self._max_s2_antes)
        self.d['blanco_paso'].append(step)
        self.d['dq'].append(dq.squeeze(1).cpu().numpy())
        self.d['dy'].append(dy.squeeze(1).cpu().numpy())

    def periodico(self, step):
        if step % self.LOG_CADA == 0 and self.n_acum:
            v = (self.acum / self.n_acum).tolist()
            for k, x in zip(['perdida', 'td_abs', 'td_mayor_1', 'grad', 'grad_recorte', 'q_sa'], v):
                self.d[k].append(x)
            self.d['log_paso'].append(step)
            self.acum.zero_()
            self.n_acum = 0
        if step % self.SONDA_CADA == 0:
            self._sondas(step)
            self._lote_tipico(step)

    def _sondas(self, step):
        """Q medio en los estados de referencia; en Dueling también V(s) y la dispersión de A(s, a)."""
        with torch.no_grad():
            h = self.net.trunk(self.sondas)
            if hasattr(self.net, 'advantage'):
                v, adv = self.net.value(h), self.net.advantage(h)
                q = v + adv - adv.mean(dim=1, keepdim=True)
            else:
                q = self.net.fc(h)
                v, adv = q.mean(dim=1, keepdim=True), q   # en DQN, la media de Q y Q mismo
        self.d['sonda_paso'].append(step)
        self.d['q_max'].append(q.max(dim=1)[0].mean().item())
        self.d['v'].append(v.mean().item())
        self.d['disp_a'].append((adv.max(dim=1)[0] - adv.min(dim=1)[0]).mean().item())

    def _lote_tipico(self, step):
        """Un lote de 32 muestras uniformes del buffer, como las de sample: antigüedad, de cuántos
        episodios distintos vienen, qué fracción es del episodio en curso y correlación entre estados."""
        b = self.buffer
        if len(b) < BATCH_SIZE:
            return
        idx = self.rng.integers(0, len(b), size=BATCH_SIZE)
        edad = (b.pos - 1 - idx) % b.capacity              # pasos desde que se guardó cada transición
        paso_guardado = step - edad
        episodio = np.searchsorted(np.array(self.fin_eps), paso_guardado)
        x = b.s[idx, -1].reshape(BATCH_SIZE, -1).astype(np.float32)
        c = np.corrcoef(x)
        self.d['lote_paso'].append(step)
        self.d['lote_edad'].append(edad.mean())
        self.d['lote_episodios'].append(len(np.unique(episodio)))
        self.d['lote_mismo_ep'].append(np.mean(paso_guardado >= self.inicio_ep))
        self.d['lote_corr'].append(np.nanmean(c[np.triu_indices(BATCH_SIZE, 1)]))

    def episodio(self, step, ep_return, epsilon):
        for k, x in [('ep_paso', step), ('ep_retorno', ep_return), ('ep_epsilon', epsilon),
                     ('ep_pasados', self.pasados), ('ep_sobrepasado', self.sobrepasado),
                     ('ep_acciones', self.acciones.copy())]:
            self.d[k].append(x)
        self.acciones[:] = 0
        self.pasados = self.sobrepasado = 0
        self.fin_eps.append(step)
        self.inicio_ep = step + 1

    def resultado(self):
        r = {k: np.array(v) for k, v in self.d.items() if k not in ('dq', 'dy')}
        r['dq'] = np.concatenate(self.d['dq']) if self.d['dq'] else np.zeros(0)
        r['dy'] = np.concatenate(self.d['dy']) if self.d['dy'] else np.zeros(0)
        return r


# ============================================================
# Evaluación y ejecución en paralelo
# ============================================================
def evaluar_episodios(net, n_eval=N_EVAL, max_steps=10_000):
    """Igual que evaluate (mismas semillas 10_000 + i y EVAL_EPSILON), pero devuelve el retorno
    y la duración de cada episodio. Con n_eval = 3 son los episodios de evaluate."""
    eval_env = make_env()
    scores, largos = [], []
    for i in range(n_eval):
        state, _ = eval_env.reset(seed=10_000 + i)
        done, total, steps = False, 0.0, 0
        while not done and steps < max_steps:
            if random.random() < EVAL_EPSILON:
                action = random.randrange(N_ACTIONS)
            else:
                with torch.no_grad():
                    s_t = torch.as_tensor(np.array(state, dtype=np.uint8), device=device).unsqueeze(0)
                    action = int(net(s_t).argmax(dim=1).item())
            state, reward, terminated, truncated, _ = eval_env.step(action)
            done = terminated or truncated
            total += reward
            steps += 1
        scores.append(total)
        largos.append(steps)
    eval_env.close()
    return np.array(scores), np.array(largos)


def _corrida(train_fn, network_class, seed, hp, n_eval, guardar_red):
    """Una corrida completa en un proceso aparte; devuelve curvas, diagnósticos y la red en CPU."""
    import ale_py
    gym.register_envs(ale_py)
    torch.set_num_threads(1)
    g = train_fn.__globals__
    previos = {k: g[k] for k in hp if k in g}
    g.update(hp)                             # hiperparámetros de esta corrida
    try:
        t0 = time.time()
        salida = train_fn(network_class, seed=seed, verbose=False)
        segundos = time.time() - t0
    finally:
        # con n_jobs = 1 joblib corre en el proceso principal: se restauran sus valores
        g.update(previos)
        for k in set(hp) - set(previos):
            del g[k]
    net, steps, returns = salida[:3]
    puntajes, largos = evaluar_episodios(net, n_eval)
    memoria = psutil.Process().memory_info()
    r = {'arq': network_class.__name__, 'semilla': seed, 'steps': steps, 'returns': returns,
         'diag': salida[3] if len(salida) > 3 else None, 'eval': puntajes, 'eval_largo': largos,
         'segundos': segundos, 'ram_gb': getattr(memoria, 'peak_wset', memoria.rss) / 1e9}
    if guardar_red:
        r['red'] = {k: v.cpu() for k, v in net.state_dict().items()}
    del net
    torch.cuda.empty_cache()
    return r


def ram_por_proceso(buffer_size):
    """s y s2 del buffer más lo que ocupan torch, CUDA y el entorno en cada proceso."""
    return 2 * buffer_size * 4 * 84 * 84 + RAM_BASE


def procesos_para(buffer_size, n_tareas):
    """Procesos simultáneos: hasta MAX_PROCESOS y solo los que entran en la RAM libre."""
    libre = psutil.virtual_memory().available - MARGEN_RAM
    return max(1, min(MAX_PROCESOS, n_tareas, int(libre // ram_por_proceso(buffer_size))))


def entrenar_configuraciones(train_fn, configuraciones, semillas=SEMILLAS, arquitecturas=ARQUITECTURAS,
                             n_eval=N_EVAL, guardar_red=False):
    """Entrena cada configuración (nombre -> cambios respecto de los hiperparámetros actuales) con cada
    arquitectura y semilla, en procesos que comparten la GPU. Devuelve nombre -> arquitectura -> corridas."""
    base = {k: globals()[k] for k in HIPERPARAMETROS if k in globals()}
    tareas = [(nombre, cls, s, {**base, **cambios}) for nombre, cambios in configuraciones.items()
              for cls in arquitecturas for s in semillas]
    get_reusable_executor().shutdown(wait=True)   # libera la RAM y la GPU de la tanda anterior
    n_jobs = procesos_para(max(hp['BUFFER_SIZE'] for *_, hp in tareas), len(tareas))
    t0 = time.time()
    salidas = Parallel(n_jobs=n_jobs, backend='loky')(
        delayed(_corrida)(train_fn, cls, s, hp, n_eval, guardar_red) for _, cls, s, hp in tareas)
    minutos = (time.time() - t0) / 60
    pasos = sum(hp['NUM_STEPS'] for *_, hp in tareas)
    print(f"{len(tareas)} corridas, {pasos / 1e6:.1f} millones de pasos en {minutos:.1f} min con {n_jobs} "
          f"procesos en paralelo ({pasos / minutos / 60:.0f} pasos/s en total); RAM máxima por proceso "
          f"{max(r['ram_gb'] for r in salidas):.1f} GB")
    res = {}
    for (nombre, cls, s, _), r in zip(tareas, salidas):
        res.setdefault(nombre, {}).setdefault(cls.__name__, []).append(r)
    return res


# Medición: pasos por segundo en total con 1, 2, 4 y 8 procesos que comparten la GPU (entrenamiento
# corto con el código original, red objetivo cada 2500 pasos y epsilon al mínimo a los 500 pasos)
MAX_PROCESOS, RAM_BASE, MARGEN_RAM = 8, 2.0e9, 6e9
prueba = dict(NUM_STEPS=3_000, TARGET_UPDATE_EVERY=2_500, EPSILON_DECAY_STEPS=500)
print("\nEntrenamientos simultáneos en la GPU (3000 pasos cada uno):")
for n in (1, 2, 4, 8):
    get_reusable_executor().shutdown(wait=True)
    hp = {**{k: globals()[k] for k in HIPERPARAMETROS if k in globals()}, **prueba}
    Parallel(n_jobs=n, backend='loky')(delayed(_corrida)(train_agent, DQN, 0, hp, 0, False) for _ in range(n))
    t0 = time.time()
    salidas = Parallel(n_jobs=n, backend='loky')(
        delayed(_corrida)(train_agent, DQN, s, hp, 0, False) for s in range(n))
    total = n * prueba['NUM_STEPS'] / (time.time() - t0)
    print(f"  {n} procesos: {total:5.0f} pasos/s en total, {total / n:4.0f} por proceso; "
          f"RAM máxima por proceso {max(r['ram_gb'] for r in salidas):.1f} GB")
print(f"Se usan hasta {MAX_PROCESOS} procesos, limitados por la RAM libre: con BUFFER_SIZE = 50_000 cada uno "
      f"necesita unos {ram_por_proceso(50_000) / 1e9:.1f} GB, así que entran "
      f"{procesos_para(50_000, 99)} procesos; con BUFFER_SIZE = 1_000, {procesos_para(1_000, 99)}.")


# ============================================================
# Tablas y curvas
# ============================================================
def media_ee(x):
    """Media y error estándar de la media."""
    x = np.asarray(x, dtype=float)
    return x.mean(), (x.std(ddof=1) / np.sqrt(len(x)) if len(x) > 1 else 0.0)


def fmt_me(x, ancho=15, dec=1):
    m, ee = media_ee(x)
    return f"{m:.{dec}f} ± {ee:.{dec}f}".rjust(ancho)


def resumir(res, etiquetas=None):
    """Por configuración y arquitectura: evaluación final (media de N_EVAL episodios con ε = 0.01) de
    cada semilla y su media entre semillas ± ee, retorno medio de los últimos 20 episodios de
    entrenamiento, Q medio en los estados de referencia al final y minutos por corrida."""
    n = max(len(c) for arqs in res.values() for c in arqs.values())
    print(f"{'configuración':<30}{'arquitectura':<13}" + ''.join(f"{f'sem {s}':>8}" for s in range(n))
          + f"{'media ± ee':>15}{'últ. 20 ep':>12}{'Q sondas':>10}{'min':>6}")
    for nombre, arqs in res.items():
        for arq, corridas in arqs.items():
            ev = [c['eval'].mean() for c in corridas]
            fila = f"{(etiquetas or {}).get(nombre, nombre):<30}{NOMBRE[arq]:<13}" + ''.join(f"{v:>8.1f}" for v in ev)
            fila += ' ' * 8 * (n - len(ev)) + fmt_me(ev, 15)
            fila += f"{np.mean([c['returns'][-20:].mean() for c in corridas]):>12.1f}"
            fila += f"{np.mean([c['diag']['q_max'][-1] for c in corridas]):>10.3g}"
            fila += f"{np.mean([c['segundos'] for c in corridas]) / 60:>6.1f}"
            print(fila)


def curva(c, grilla, w=5):
    """Media móvil de w episodios (como moving_average) interpolada en una grilla de pasos."""
    ma = moving_average(c['returns'], w)
    x = c['steps'][len(c['steps']) - len(ma):]
    return np.where((grilla >= x[0]) & (grilla <= x[-1]), np.interp(grilla, x, ma), np.nan)


def graficar_curvas(ax, corridas, color, etiqueta, num_steps, finas=True):
    """Media entre semillas ± desvío de la curva de cada semilla (líneas finas)."""
    grilla = np.linspace(0, num_steps, 201)
    Y = np.array([curva(c, grilla) for c in corridas])
    if finas:
        for y in Y:
            ax.plot(grilla, y, color=color, linewidth=0.7, alpha=0.35)
    m, s = np.nanmean(Y, axis=0), np.nanstd(Y, axis=0)
    ax.plot(grilla, m, color=color, linewidth=2.2, label=etiqueta)
    ax.fill_between(grilla, m - s, m + s, color=color, alpha=0.15)
    ax.set_xlim(0, num_steps)


def retorno_aleatorio(n_eval=N_EVAL):
    """Política al azar en los mismos episodios de evaluación."""
    env = make_env()
    rng = np.random.default_rng(0)
    scores = []
    for i in range(n_eval):
        env.reset(seed=10_000 + i)
        done, total = False, 0.0
        while not done:
            _, reward, term, trunc, _ = env.step(int(rng.integers(N_ACTIONS)))
            done = term or trunc
            total += reward
        scores.append(total)
    env.close()
    return np.array(scores)


RET_ALEATORIO = retorno_aleatorio()
print(f"\nPolítica al azar en los {N_EVAL} episodios de evaluación: retorno medio {RET_ALEATORIO.mean():.1f} "
      f"(mínimo {RET_ALEATORIO.min():.0f}, máximo {RET_ALEATORIO.max():.0f})")
```

```python
# Pregunta 1: cambie TARGET_UPDATE_EVERY a 2_500, vuelva a entrenar ambas arquitecturas y compare las curvas
# Código original de la celda 5 con el cambio de la pregunta 1. La lógica del entrenamiento no cambia:
# solo se agrega el registro de diagnósticos (Registro, definido en la preparación), que se devuelve.
TARGET_UPDATE_EVERY = 2_500   # Pregunta 1: la red objetivo se copia cada 2500 pasos (antes, en cada paso)


def train_agent(network_class, seed=SEED, num_steps=None, verbose=True):
    num_steps = num_steps or NUM_STEPS
    env = make_env()
    random.seed(seed); np.random.seed(seed); torch.manual_seed(seed)

    policy_net = network_class(N_ACTIONS).to(device)
    target_net = network_class(N_ACTIONS).to(device)
    target_net.load_state_dict(policy_net.state_dict())
    optimizer = torch.optim.Adam(policy_net.parameters(), lr=LR)
    buffer = ReplayBuffer(BUFFER_SIZE)
    registro = Registro(policy_net, buffer, seed)   # diagnósticos para las respuestas

    state, _ = env.reset(seed=seed)
    state = np.array(state, dtype=np.uint8)
    ep_return = 0.0
    log_steps, log_returns = [], []
    t0 = time.time()

    for step in range(1, num_steps + 1):
        # Política de comportamiento epsilon-greedy con decaimiento lineal
        epsilon = max(EPSILON_END,
                      EPSILON_START - (EPSILON_START - EPSILON_END) * step / EPSILON_DECAY_STEPS)
        if random.random() < epsilon:
            action = random.randrange(N_ACTIONS)
        else:
            with torch.no_grad():
                s_t = torch.as_tensor(state, device=device).unsqueeze(0)
                action = int(policy_net(s_t).argmax(dim=1).item())

        next_state, reward, terminated, truncated, _ = env.step(action)
        next_state = np.array(next_state, dtype=np.uint8)
        done = terminated or truncated
        # Solo 'terminated' corta el bootstrap (truncated no es un estado terminal real)
        buffer.push(state, action, reward, next_state, float(terminated))
        state = next_state
        ep_return += reward
        registro.paso(action, reward)

        if len(buffer) >= max(BATCH_SIZE, LEARN_START) and step % TRAIN_EVERY == 0:
            s, a, r, s2, d = buffer.sample(BATCH_SIZE)
            q_sa = policy_net(s).gather(1, a)
            with torch.no_grad():
                q_next = target_net(s2).max(dim=1, keepdim=True)[0]
                target = r + GAMMA * q_next * (1.0 - d)
            loss = F.mse_loss(q_sa, target)
            optimizer.zero_grad()
            loss.backward()
            registro.antes_del_paso(loss, q_sa, target, s2, d)
            optimizer.step()
            registro.despues_del_paso(step, s, a, s2, d)

        # Pregunta 1: copia periódica de la red principal a la red objetivo
        if step % TARGET_UPDATE_EVERY == 0:
            target_net.load_state_dict(policy_net.state_dict())
        registro.periodico(step)

        if done:
            log_steps.append(step)
            log_returns.append(ep_return)
            registro.episodio(step, ep_return, epsilon)
            if verbose:
                print(f"  Paso {step:8d} | ε={epsilon:.3f} | Episodio {len(log_returns):3d} | "
                      f"Retorno: {ep_return:6.1f} | Media (últ. 5): {np.mean(log_returns[-5:]):6.1f}")
            ep_return = 0.0
            state, _ = env.reset()
            state = np.array(state, dtype=np.uint8)

    if verbose:
        print(f"  Tiempo de entrenamiento: {(time.time() - t0) / 60:.1f} min")
    env.close()
    del buffer
    gc.collect()
    if device.type == "cuda":
        torch.cuda.empty_cache()
    return policy_net, np.array(log_steps), np.array(log_returns), registro.resultado()


# Se entrenan las dos arquitecturas con 3 semillas, con la red objetivo original (copia en cada paso)
# y con la de la pregunta 1, en la misma tanda de procesos
res_p1 = entrenar_configuraciones(train_agent, {'original': dict(TARGET_UPDATE_EVERY=1), 'p1': {}})
ETIQ = {'original': 'objetivo en cada paso', 'p1': 'objetivo cada 2500 pasos'}
print()
resumir(res_p1, ETIQ)


def pendiente_blanco(c):
    """Por cada unidad que sube Q(s, a) en una actualización, cuánto sube r + γ max Q(s′, ·) con la misma red."""
    dq, dy = c['diag']['dq'], c['diag']['dy']
    return np.sum(dq * dy) / np.sum(dq * dq)


def fase(c, clave, desde, hasta):
    """Media de un diagnóstico registrado cada LOG_CADA pasos, entre dos pasos."""
    x, y = c['diag']['log_paso'], c['diag'][clave]
    return y[(x > desde) & (x <= hasta)].mean()


print("\nDiagnósticos (media de 3 semillas): Q sondas es max_a Q(s, a) medio en los 512 estados de referencia")
print(f"{'configuración':<30}{'arquitectura':<13}{'arrastre':>10}{'Q sondas':>10}{'máx |Q|':>10}"
      f"{'oscil. Q':>10}{'pérdida':>10}{'|δ|':>9}{'norma grad':>12}{'acelera':>9}{'pasados':>9}")
for nombre, arqs in res_p1.items():
    for arq, corridas in arqs.items():
        qs = [c['diag']['q_max'] for c in corridas]
        acc = [c['diag']['ep_acciones'][-10:] for c in corridas]
        print(f"{ETIQ[nombre]:<30}{NOMBRE[arq]:<13}"
              f"{np.mean([pendiente_blanco(c) for c in corridas]):>10.2f}"
              f"{np.mean([q[-1] for q in qs]):>10.3g}{np.mean([np.abs(q).max() for q in qs]):>10.3g}"
              f"{np.mean([np.abs(np.diff(q)).mean() for q in qs]):>10.3g}"
              f"{np.mean([fase(c, 'perdida', 50_000, NUM_STEPS) for c in corridas]):>10.3g}"
              f"{np.mean([fase(c, 'td_abs', 50_000, NUM_STEPS) for c in corridas]):>9.3g}"
              f"{np.mean([fase(c, 'grad', 50_000, NUM_STEPS) for c in corridas]):>12.3g}"
              f"{np.mean([a[:, ACELERA].sum() / a.sum() for a in acc]):>9.2f}"
              f"{np.mean([c['diag']['ep_pasados'][-10:].mean() for c in corridas]):>9.1f}")
print("arrastre: pendiente de la regresión de Δ[r + γ max Q(s′, ·)] sobre ΔQ(s, a) en cada actualización")
print("oscil. Q: cambio medio de Q sondas entre mediciones (cada 5000 pasos); pérdida, |δ| y norma del")
print("gradiente: media desde el paso 50000; acelera y pasados: fracción de acciones con FIRE y autos")
print("pasados por episodio en los últimos 10 episodios")

# Gráficos: curvas de aprendizaje, Q en los estados de referencia y error TD
fig, axs = plt.subplots(1, 3, figsize=(18, 5))
estilos = {'original': ':', 'p1': '-'}
for nombre, arqs in res_p1.items():
    for arq, corridas in arqs.items():
        etiqueta = f"{NOMBRE[arq]}, {ETIQ[nombre]}"
        graficar_curvas(axs[0], corridas, COLOR_ARQ[arq], etiqueta, NUM_STEPS, finas=False)
        axs[0].lines[-1].set_linestyle(estilos[nombre])
        for i, c in enumerate(corridas):
            d = c['diag']
            axs[1].plot(d['sonda_paso'], d['q_max'], estilos[nombre], color=COLOR_ARQ[arq], linewidth=1.3,
                        alpha=0.85, label=etiqueta if i == 0 else None)
            axs[2].plot(d['log_paso'], d['td_abs'], estilos[nombre], color=COLOR_ARQ[arq], linewidth=1,
                        alpha=0.7, label=etiqueta if i == 0 else None)
axs[0].set_title('Pregunta 1: retorno (media de 3 semillas ± desvío)')
axs[0].set_ylabel('Retorno por episodio (media móvil de 5)')
axs[1].set_title('Q medio en los estados de referencia (cada semilla)')
axs[1].set_ylabel('max_a Q(s, a) medio (escala simétrica logarítmica)')
axs[1].set_yscale('symlog', linthresh=1)
axs[2].set_title('Error TD medio |r + γ max Q_obj(s′, ·) − Q(s, a)|')
axs[2].set_ylabel('|δ| (media cada 1000 pasos)')
axs[2].set_yscale('log')
for ax in axs:
    ax.set_xlabel('Pasos de entorno')
    ax.grid(alpha=0.2)
    ax.legend(fontsize=8)
plt.tight_layout()
plt.show()
```

Por qué un error TD grande es especialmente dañino con la pérdida cuadrática. El gradiente de (y − Q)² respecto de Q es −2(y − Q), proporcional al error: una muestra con δ = 100 empuja 100 veces más que una con δ = 1 y en la pérdida pesa 10000 veces más, así que un solo error grande (una transición rara, o un objetivo sobreestimado) domina el lote. Como los pesos son compartidos, ese salto cambia Q en todos los estados, también en los s′ de los objetivos, y puede generar errores todavía mayores; es la realimentación que hizo divergir a la DQN de la pregunta 1, con un error TD medio de 1.2·10⁷. Relación con la ventaja de la arquitectura Dueling en Enduro. La mayor parte del tiempo elegir una u otra acción cambia poco y lo que importa es la situación: el momento del día, la velocidad, si acaba de chocar. En DQN cada actualización corrige solo el Q(s, a) de la acción tomada, y las otras 8 acciones aprenden el valor del estado solo a través de las capas compartidas. En Dueling DQN la rama de V es común a las 9 acciones, así que cada transición actualiza el valor del estado sin importar qué acción se tomó, y la rama de A solo tiene que aprender las diferencias chicas y los pocos momentos críticos, como un auto justo adelante. Es lo mismo que muestra el paper en Enduro: V presta atención a los autos lejanos, donde empiezan a verse, y al puntaje, y A solo se activa cuando hay un auto cerca adelante que obliga a elegir bien la acción.

Huber bajó la norma del gradiente, el recorte no se activó nunca (la norma nunca superó 10) y LR = 10⁻⁴ retrasó el arranque.
